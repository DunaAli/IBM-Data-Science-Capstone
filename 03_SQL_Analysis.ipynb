# SpaceX Falcon 9 — SQL Analysis

The IBM prepared CSV is loaded into an in-memory SQLite database, then queried using SQL.

```python
import pandas as pd
import sqlite3
df=pd.read_csv('dataset_part_2.csv')
db=sqlite3.connect(':memory:')
df.to_sql('spacex',db,index=False,if_exists='replace')
```

```python
queries={
'Launch sites': 'SELECT LaunchSite, COUNT(*) AS launches, SUM(Class) AS successes, ROUND(100.0*AVG(Class),1) AS success_pct FROM spacex GROUP BY LaunchSite ORDER BY launches DESC',
'Orbits': 'SELECT Orbit, COUNT(*) AS launches, ROUND(100.0*AVG(Class),1) AS success_pct FROM spacex GROUP BY Orbit ORDER BY launches DESC',
'Outcomes': 'SELECT Outcome, COUNT(*) AS launches FROM spacex GROUP BY Outcome ORDER BY launches DESC',
'Payload bands': "SELECT CASE WHEN PayloadMass < 4000 THEN 'Under 4000 kg' WHEN PayloadMass < 10000 THEN '4000-9999 kg' ELSE '10000+ kg' END AS payload_band, COUNT(*) AS launches, ROUND(100.0*AVG(Class),1) AS success_pct FROM spacex GROUP BY payload_band ORDER BY launches DESC",
'Annual trend': "SELECT SUBSTR(Date,1,4) AS launch_year, COUNT(*) AS launches, ROUND(100.0*AVG(Class),1) AS success_pct FROM spacex GROUP BY launch_year ORDER BY launch_year"
}
for label,query in queries.items():
    print('\n'+label)
    display(pd.read_sql_query(query,db))
```
