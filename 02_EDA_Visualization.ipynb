# SpaceX Falcon 9 — Exploratory Data Analysis

Plots are reproduced from the prepared IBM course dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('dataset_part_2.csv')
df['Year'] = pd.to_datetime(df['Date']).dt.year
```

```python
site = df.groupby('LaunchSite')['Class'].agg(['count','sum'])
site.plot(kind='bar', figsize=(9,4), title='Launches and successful landings by site')
plt.ylabel('Launches'); plt.tight_layout(); plt.show()
```

```python
for outcome,label in [(0,'Not successful'),(1,'Successful')]:
    subset=df[df['Class']==outcome]
    plt.scatter(subset['PayloadMass'],subset['FlightNumber'],label=label,alpha=.7)
plt.xlabel('Payload mass (kg)'); plt.ylabel('Flight number')
plt.legend(); plt.title('Payload mass, flight number and landing result')
plt.tight_layout(); plt.show()
```

```python
annual=df.groupby('Year')['Class'].mean()*100
annual.plot(marker='o',figsize=(9,4),title='Landing success rate by year')
plt.ylabel('Success rate (%)'); plt.ylim(0,105); plt.tight_layout(); plt.show()
```

```python
orbit=df.groupby('Orbit')['Class'].agg(['count','mean'])
orbit[orbit['count']>=3]['mean'].sort_values().mul(100).plot(kind='bar',figsize=(9,4))
plt.ylabel('Success rate (%)'); plt.title('Orbit success rate (3+ missions)')
plt.tight_layout(); plt.show()
```
