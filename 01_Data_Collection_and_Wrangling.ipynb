# SpaceX Falcon 9 — Dataset and Data Wrangling

This notebook **loads the IBM Skills Network prepared CSV**. It does not claim to have run the original live SpaceX API or Wikipedia scraping exercises.

```python
import pandas as pd
df = pd.read_csv('dataset_part_2.csv')
print('Rows, columns:', df.shape)
display(df.head())
```

```python
print(df.isna().sum().sort_values(ascending=False).head(10))
print('Target distribution:')
print(df['Class'].value_counts().sort_index())
print('Unique launch sites:', df['LaunchSite'].nunique())
```

```python
df['Year'] = pd.to_datetime(df['Date']).dt.year
print('Success by launch site')
display(df.groupby('LaunchSite')['Class'].agg(['count','sum','mean']))
print('Success by year')
display(df.groupby('Year')['Class'].agg(['count','mean']))
```
