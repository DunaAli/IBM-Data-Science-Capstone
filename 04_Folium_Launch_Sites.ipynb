# SpaceX Falcon 9 — Folium Launch-Site Map

This notebook maps launch-site coordinates from the prepared IBM CSV. **Coastline/road/railway proximity distances have not been computed**; the map does not claim otherwise.

```python
import pandas as pd
import folium
df=pd.read_csv('dataset_part_2.csv')
sites=df.groupby('LaunchSite').agg(Latitude=('Latitude','first'),Longitude=('Longitude','first'),Launches=('Class','size'),Successes=('Class','sum')).reset_index()
display(sites)
```

```python
m=folium.Map(location=[sites.Latitude.mean(),sites.Longitude.mean()],zoom_start=4)
for _,r in sites.iterrows():
    folium.Marker([r.Latitude,r.Longitude],popup=f"{r.LaunchSite}: {r.Launches} launches, {r.Successes} successes",tooltip=r.LaunchSite).add_to(m)
m.save('SpaceX_Folium_Actual_Map.html')
m
```
